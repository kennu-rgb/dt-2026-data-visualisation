In this task, I will use a text file called "Short Stories.txt" to create a word cloud. I will analyze the text and identify the most frequently used words. I printed in total 100 000 characters of text.

```python
file = open("Short Stories.txt", "r", encoding="utf-8")
text = file.read()
file.close()

print(text[:100000])
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

stopwords_list = [
    "the", "and", "was", "of", "to", "a", "in", "is",
    "it", "he", "she", "they", "his", "her", "their",
    "that", "this", "with", "for", "on", "at", "as",
    "but", "not", "be", "have", "had", "were", "are",
    "you", "i", "we", "my", "me", "him", "them",
    "all", "so", "if", "then", "there", "when",
    "what", "which", "who", "would", "could",
    "into", "out", "up", "down", "about", "from",
    "very", "just", "now", "can", "will", "shall",
    "do", "does", "did", "done", "has", "having",
    "your", "our", "their", "these", "those",
    "than", "too", "also", "being", "been",
    "no", "yes", "or", "nor", "because", "while",
    "after", "before", "again", "here", "where",
    "how", "any", "each", "both", "few", "own", "by", "am", "an"
    ,"know", "only","why" ]

wc = WordCloud(
    width=800,
    height=400,
    background_color="white",
    stopwords=stopwords_list
).generate(text)

plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.show()
```

The word cloud shows the most frequently used words in the short stories. Larger words appear more often in the text. It helps identify the main topics and ideas in the stories. I removed common words using stopwords to make the word cloud clearer.
