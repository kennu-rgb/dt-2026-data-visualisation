```python
# can use to work with tabular data
import pandas as pd
df = pd.read_csv("competition.csv")
df.head() # Show the first few rows
df.describe() #Show some basic statistics
```

```python
import matplotlib.pyplot as plt #matplotlib helps to create charts and visualizations
df.plot(kind="bar", x="Name", y="Years_of_practice", legend=False)
plt.title("Years of practice per Athlete")
plt.ylabel("Years of practice")
plt.show()
```

```python
import matplotlib.pyplot as plt #matplotlib helps to create charts and visualizations
df.plot(kind="bar", x="Name", y="Age", legend=False)
plt.title("Ages of the athletes")
plt.ylabel("Age")
plt.show()
```

```python
df.plot(kind="pie", y="Age", labels=df["Name"], legend=False, autopct="%1.1f%%")
plt.title("Athletes' ages as percentages of the total")
```

```python
plt.scatter(df["Age"], df["Years_of_practice"], s=df["Years_of_practice"]*60) 
#s=df["Years_of_practice"]*60 this means if students have more years of practice their bubble is bigger
#alpha=0.6 makes the bubbles a bit transparent
plt.title("Ages compared to years of practice")
plt.ylabel("Years of practce")
plt.xlabel("Age")
```

```python
import seaborn as sns #use this to create more clean and formatted plots
sns.boxplot(data=df, y="Age")
plt.title("Distribution of athlete age")
plt.show()
```

```python
plt.Figure(figsize=(8,5))
df[["Age", "Years_of_practice"]].boxplot()
plt.title("Box plot for age and years of practice")
plt.ylabel("Years")
plt.show
```

```python
avg_age= df.groupby("Sport")["Age"].mean()
avg_age.plot(kind="bar")
plt.show()
```

```python
avg_Years_of_practice= df.groupby("Sport")["Years_of_practice"].mean()
avg_Years_of_practice.plot(kind="bar")
plt.title("Average years of practice per sport")
plt.ylabel("Years of practice")
plt.show()
```
