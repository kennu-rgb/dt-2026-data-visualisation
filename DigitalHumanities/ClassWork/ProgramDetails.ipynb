```python
import pandas as pd
df = pd.read_csv ("Competencies.csv")
df
```

```python
import matplotlib.pyplot as plt
```

Chart showing who puts in the most effort

```python
avg_Hours_Studied= df.groupby("Student")["Hours_Studied"].mean()
avg_Hours_Studied.plot(kind="bar",color="#FFC0CB")
plt.title("How many hours do the students study on average?")
plt.ylabel("Hours on average")
plt.show()
```

```python
plt.scatter (df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30, alpha=0.6)
plt.ylabel("Grade")
plt.xlabel("Hours Studied")
plt.show()
```

```python
plt.Figure(figsize=(8,5))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("Box plot for hours studied and grade")
plt.show()
```
