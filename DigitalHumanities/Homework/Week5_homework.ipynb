```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../Homework/italian_regions_cultural_heritage.csv")
df.head()
```

```python
plt.bar(df["Region"], df["UNESCO_Sites"])
plt.xticks(rotation=45)
plt.xlabel("Region")
plt.ylabel("UNESCO Sites")
plt.title("UNESCO Sites by Italian Region")
plt.show()
```

The chart shows that the number of UNESCO sites differs between the Italian regions. Lombardy has the highest number of UNESCO sites in this dataset, while Emilia-Romagna has the lowest.

```python
plt.scatter(df["Population_millions"], df["UNESCO_Sites"])
plt.xlabel("Population (millions)")
plt.ylabel("UNESCO Sites")
plt.title("Population vs. UNESCO Sites")
plt.show()
```

The scatter plot shows that there is not a clear relationship between population and the number of UNESCO sites. Lombardy has both the highest population and the highest number of UNESCO sites, but other regions with smaller populations also have several UNESCO sites.

```python
import matplotlib.pyplot as plt

sorted_df = df.sort_values("Area_km2")

plt.bar(sorted_df["Region"], sorted_df["Area_km2"])

plt.title("Area of Italian Regions")
plt.xlabel("Region")
plt.ylabel("Area (km²)")

plt.xticks(rotation=45)
plt.show()
```

The chart shows that the Italian regions have different geographical areas. Campania has the smallest area, while Sicily has the largest area in this dataset. This shows that the size of a region does not necessarily determine how many UNESCO sites it has.
